## 1. Conditional Statements
Python uses conditional statements to execute code blocks selectively based on boolean logic.

```python
# Simple conditional check
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
else:
    grade = 'C'

print(f"Score: {score}, Grade: {grade}")
```

## 2. Loops
Loops allow us to repeat blocks of code. We use `for` loops for definite iteration and `while` loops for indefinite iteration.

```python
# Iterating over a list using a for loop
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(f"I like {fruit}")
```

```python
# Using a while loop with a counter
count = 3
while count > 0:
    print(f"Countdown: {count}")
    count -= 1
print("Blast off!")
```

## 3. Loop Control: Break and Continue
- `continue` skips the current iteration.
- `break` exits the loop immediately.

```python
# Demonstration of break and continue
numbers = [1, -2, 3, 999, 4, 5]

for num in numbers:
    if num < 0:
        print("Skipping negative number")
        continue  # Skip to next iteration
    if num == 999:
        print("Found stop signal (999). Stopping loop.")
        break  # Exit loop completely
    print(f"Processing: {num}")
```
